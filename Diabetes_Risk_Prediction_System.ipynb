# Diabetes Risk Prediction System

**Internship Project — Machine Learning Classification**

This project uses the provided PIMA diabetes dataset to build a binary classification model with a linear Support Vector Machine (SVM). The workflow includes data analysis, standardization, stratified train/test splitting, model training, evaluation, and a sample prediction.

## 1. Import Dependencies

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

## 2. Load the Dataset

The supplied `diabetes.csv` file contains 768 records and 9 columns. `Outcome` is the target: 0 = non-diabetic and 1 = diabetic, following the original notebook's labeling.

```python
diabetes_dataset = pd.read_csv("diabetes.csv")
diabetes_dataset.head()
```

```python
print("Dataset shape:", diabetes_dataset.shape)
print("\nColumns:", list(diabetes_dataset.columns))
```

```python
diabetes_dataset.describe()
```

```python
print(diabetes_dataset["Outcome"].value_counts())
```

## 3. Separate Features and Target

```python
X = diabetes_dataset.drop(columns="Outcome", axis=1)
Y = diabetes_dataset["Outcome"]

print("Feature shape:", X.shape)
print("Target shape:", Y.shape)
```

## 4. Data Standardization

Standardization puts the numerical features on a comparable scale before SVM training.

```python
scaler = StandardScaler()
scaler.fit(X)

standardized_data = scaler.transform(X)
X = standardized_data
Y = diabetes_dataset["Outcome"]
```

## 5. Train-Test Split

A stratified 80/20 split is used with `random_state=2`, matching the original project workflow.

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, stratify=Y, random_state=2
)

print("Full dataset:", X.shape)
print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)
```

## 6. Train the SVM Classifier

```python
classifier = svm.SVC(kernel="linear")
classifier.fit(X_train, Y_train)
```

## 7. Model Evaluation

```python
X_train_prediction = classifier.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction, Y_train)

X_test_prediction = classifier.predict(X_test)
test_data_accuracy = accuracy_score(X_test_prediction, Y_test)

print("Training accuracy:", training_data_accuracy)
print("Test accuracy:", test_data_accuracy)
```

```python
print("Confusion matrix:")
print(confusion_matrix(Y_test, X_test_prediction))

print("\nClassification report:")
print(classification_report(Y_test, X_test_prediction))
```

## 8. Sample Predictive System

The following example uses the same sample input from the original notebook.

```python
input_data = (5, 166, 72, 19, 175, 25.8, 0.587, 51)

input_data_as_numpy_array = np.asarray(input_data)
input_data_reshaped = input_data_as_numpy_array.reshape(1, -1)

std_data = scaler.transform(input_data_reshaped)
prediction = classifier.predict(std_data)

if prediction[0] == 0:
    print("Prediction: The person is classified as non-diabetic")
else:
    print("Prediction: The person is classified as diabetic")
```
